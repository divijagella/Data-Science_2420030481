```python
import numpy as np
from scipy.spatial import distance
pointA=np.array([2,4,6])
pointB=np.array([3,6,9])
euclidean_distance=distance.euclidean(pointA,pointB)
print("Euclidean Distance",euclidean_distance)
similarity_distance=1/(1+euclidean_distance)
print("Similarity Distance",similarity_distance)
```

```python
import numpy as np
from scipy.spatial import distance
pointA=np.array([2,4,6])
pointB=np.array([3,6,9])
manhattan_distance=distance.cityblock(pointA,pointB)
print("Manhattan Distance",manhattan_distance)
similarity_manhattan=1/(1+manhattan_distance)
print("Similarity Manhattan",similarity_manhattan)
```

```python
import numpy as np
from scipy.spatial import distance
pointA=np.array([2,4,6])
pointB=np.array([3,6,9])
minkowski_distance_p3=distance.minkowski(pointA,pointB,p=3)
print("Minkowski Distance(p=3):",minkowski_distance_p3)
similarity_minkowski=1/(1+minkowski_distance_p3)
print("Minkowski Similarity(p=3):",similarity_minkowski)
```

```python
import pandas as pd
from scipy.stats import pearsonr


tips = pd.read_csv("tips.csv")


corr, p_value = pearsonr(tips['total_bill'], tips['tip'])

print(f"Pearson correlation coefficient: {corr}")
print(f"P-value: {p_value}")
```

```python
import pandas as pd
from scipy.stats import spearmanr
df=pd.DataFrame({'X':[10,20,30,40,50], 'Y':[12,18,33,47,55]})
corr_value, p_value = spearmanr(df['X'], df['Y'])
print(f"Spearman Correlation Coefficient: {corr_value}")
print(f"P-value: {p_value}")
```

```python
def hamming_distance(str1,str2):
  if len(str1)!=len(str2):
    raise ValueError("Strings must be of equal length")
    return sum(ch1!=ch2 for ch1,ch2 in zip(str1,str2))
s1="karolin"
s2="kathrin"
dist=hamming_distance(s1,s2)
print("Hamming Distance between",s1, "and",s2,"is:",dist)
```

```python
def jaccard_index(str1,str2):
  set1,set2=set(str1.split()),set(str2.split())
  intersection=set1.intersection(set2)
  union=set1.union(set2)
  return len(intersection)/len(union)
s1="data science is fun"
s2="science makes data useful"
print("Jaccard Index:",jaccard_index(s1,s2))
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
s1 = "data science is fun"
s2 = "science makes data useful"
vectorizer = CountVectorizer().fit([s1, s2])
vectors = vectorizer.transform([s1, s2])
cos_sim = cosine_similarity(vectors[0], vectors[1])[0][0]
print("Cosine Similarity:", cos_sim)
```

```python
def lcs_length(X, Y):
    m, n = len(X), len(Y)

    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m):
        for j in range(n):
            if X[i] == Y[j]:
                dp[i+1][j+1] = dp[i][j] + 1
            else:
                dp[i+1][j+1] = max(dp[i][j+1], dp[i+1][j])
    return dp[m][n]

seq1 = "ABCDEF"
seq2 = "AEBDF"
length = lcs_length(seq1, seq2)
print(f"Longest Common Subsequence length between '{seq1}' and '{seq2}': {length}")
```
